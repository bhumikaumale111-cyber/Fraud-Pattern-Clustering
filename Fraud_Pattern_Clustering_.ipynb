```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("creditcard.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
# Select features for clustering
X = df.drop(['Class'], axis=1)
```

```python
# Scale the Amount Feature
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X['Amount'] = scaler.fit_transform(X[['Amount']])
```

```python
# ELBOW METHOD TO FIND K
```

```python
from sklearn.cluster import KMeans
```

```python
X.dropna(inplace=True)
wcss = []
for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(6,4))
plt.plot(range(2, 11), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("K")
plt.ylabel("WCSS")
plt.show()
```

```python
# Apply K-Means Clustering
kmeans = KMeans(n_clusters=5, random_state=42)
clusters = kmeans.fit_predict(X)

# Align df with the processed X's index before adding the cluster column
df = df.loc[X.index].copy()
df['Cluster'] = clusters
```

```python
# Silhouette Score

from sklearn.metrics import silhouette_score

score = silhouette_score(X, clusters)
score
```

```python
# PCA for Visualization

from sklearn.decomposition import PCA
import seaborn as sns

pca = PCA(n_components=2)
pca_result = pca.fit_transform(X)

df['PC1'] = pca_result[:, 0]
df['PC2'] = pca_result[:, 1]

plt.figure(figsize=(7,5))
sns.scatterplot(data=df, x='PC1', y='PC2', hue='Cluster', palette='tab10', s=10)
plt.title("Fraud Pattern Clusters (PCA Visualization)")
plt.show()
```

```python
# Analyze Each Cluster
cluster_summary = df.groupby('Cluster').mean()
cluster_summary[['Amount']]
```

```python
# Fraud Pattern Detection

df['Cluster_Distance'] = kmeans.transform(X).min(axis=1)

plt.figure(figsize=(7,4))
sns.boxplot(x=df['Cluster_Distance'])
plt.title("Cluster Distance Distribution")
plt.show()
```

```python
# High-Risk Transaction Tagging

threshold = df['Cluster_Distance'].quantile(0.99)

df['Fraud_Risk'] = np.where(df['Cluster_Distance'] > threshold, 1, 0)

df[['Amount', 'Cluster_Distance', 'Fraud_Risk']].head(10)
```

```python
# Visualize High-Risk Transactions
plt.figure(figsize=(7,5))
sns.scatterplot(
    data=df,
    x='PC1',
    y='PC2',
    hue='Fraud_Risk',
    palette={0: 'blue', 1: 'red'},
    s=10
)
plt.title("High-Risk Transaction Visualization")
plt.show()
```

```python
## Business Insights

> Clusters show different types of spending and transaction behavior.
> Most normal transactions stay inside their cluster boundaries.
> Transactions far from their cluster center indicate unusual or risky behavior.
> Only a small percentage of transactions appear as high-risk outliers.
> Banks/Fintech can use these patterns to detect fraud early, even without labeled data.
```

```python
## Final Summary

> Built a Fraud Pattern Clustering Model using unsupervised machine learning.
> Used KMeans to find hidden patterns in transaction data.
> Applied PCA for dimensionality reduction and clear 2D visualization.
> Calculated distance-based anomaly scores to mark high-risk transactions.
> Model helps identify suspicious spending without needing fraud labels.
> Project is practical, scalable, and suitable for real financial applications.
```
